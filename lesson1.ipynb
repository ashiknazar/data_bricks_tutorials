## What does Databricks provide?

#### 1. Apache Spark

- Spark is the distributed processing engine.
  - df = spark.read.csv("/sales.csv")
  
You don't create the SparkSession; Databricks provides it automatically.

#### 2. Notebooks

Interactive notebooks that support:

- Python
- SQL
- Scala
- R


#### 3. Cluster Management

- Instead of:

  - Install Spark
  - Configure Workers
  - Edit XML files
  - Start Services

You simply choose:

- Cluster Size: 4 workers
- Runtime: 15.x
- Click Start

Databricks handles the rest.

___

### 4. Delta Lake

- Delta Lake is an optimized storage layer that adds capabilities such as ACID transactions, schema enforcement, and time travel to data stored in the lakehouse.

### 5. SQL Warehouse

Run SQL directly:
```sql
SELECT *
FROM sales
WHERE amount > 1000;
```

### 6. Job Scheduling

- Instead of using cron jobs:
- Run ETL every day
- Create a Databricks Job with a schedule.
