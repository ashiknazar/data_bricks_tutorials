### Lesson 2 — Databricks Architecture

- The most important idea is that Databricks separates storage from compute.

```tree
                    DATABRICKS
                         │
          ┌──────────────┼──────────────┐
          │              │              │
      Workspace       Compute        Data
          │              │              │
      Notebooks       Clusters       Storage
      SQL            Spark          S3 / ADLS / GCS
      Jobs
          │              │
          └──────────────┤
                         │
                    Apache Spark

```

There are three major things to understand:

- Workspace — where you work
- Compute — machines that execute your code
- Storage — where your data actually lives
